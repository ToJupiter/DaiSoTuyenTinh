```python
import matplotlib.pyplot as plt
import numpy as np
```

Những ngày tháng học trò, véc tơ trong tôi là lực kéo, vận tốc, sự dịch chuyển.

Trong đại số tuyến tính, chúng ta sẽ mở rộng góc nhìn này. Vector như một "danh sách" có thứ tự các con số, một thế giới xung quanh chúng ta đầy rẫy những "vector" không chỉ có 2 hay 3 chiều. Một bức ảnh là một vector khổng lồ của các điểm ảnh. Dữ liệu tài chính là các vector biểu diễn giá cổ phiếu theo thời gian. Giải các hệ phương trình khổng lồ, phân tích dữ liệu phức tạp, huấn luyện các mô hình học máy, vector chính là bước khởi đầu quan trọng nhất, mở cánh cửa đến một thế giới quan mà ở đó từng bước đi có dấu chân của toán học, nơi những con số ẩn chứa chút diệu kì.

**Vector: Một đoạn thẳng có hướng**

Trong toán học và vật lý, vector được hiểu là một đoạn thẳng có hướng, mang trong mình cả ý nghĩa về phương, chiều và độ lớn. Để định lượng "độ lớn" của một vector, chúng ta sử dụng một thuật ngữ đặc biệt: "chuẩn" (norm), trừ trường hợp ngoại lệ là vector không ($ \mathbf{0} $), vốn không có độ lớn.

Công thức tổng quát để xác định chuẩn của một vector trong không gian $\mathbb{R}^k$ được biểu diễn như sau:

$$
\| \mathbf{u} \| = \sqrt{x_1^2 + x_2^2 + \cdots + x_k^2}
$$

Để dễ hình dung, hãy cùng xem xét hai vector điển hình trong không gian hai chiều $\mathbb{R}^2$:

**1. Vector $(4, 7)$:**

Áp dụng công thức chuẩn, ta có:

$$
\|(4, 7)\| = \sqrt{4^2 + 7^2} = \sqrt{16 + 49} = \sqrt{65} \approx 8.06
$$

**2. Vector $(8, 6)$:**

Tương tự, chuẩn của vector này là:

$$
\|(8, 6)\| = \sqrt{8^2 + 6^2} = \sqrt{64 + 36} = \sqrt{100} = 10
$$

```python

# Define the figure and axis
fig, ax = plt.subplots(figsize=(8, 8))

# Define the vectors as starting points and directions
vectors = np.array([[0, 0, 4, 7], [0, 0, 8, 6]])

# Unpack the vectors into separate components
X, Y, U, V = zip(*vectors)

# Plot the vectors using quiver
ax.quiver(X, Y, U, V, angles='xy', scale_units='xy', scale=1, color='red', alpha=0.6)

# Set the limits and labels for the plot
ax.set_xlim([0, 10])
ax.set_ylim([0, 10])
ax.set_xlabel('Trục x', fontsize=16)
ax.set_ylabel('Trục y', fontsize=16)

# Add grid
ax.grid()

# Annotate the end points of the vectors
ax.text(4, 7, '$(4, 7)$', fontsize=16)
ax.text(8, 6, '$(8, 6)$', fontsize=16)

# Draw lines to the origin for the first vector
ax.plot([4, 4], [0, 7], color='blue', linewidth=2)
ax.plot([0, 4], [0, 0], color='blue', linewidth=2)

# Draw lines to the origin for the second vector
ax.plot([8, 0], [0, 0], color='blue', linewidth=2)
ax.plot([8, 8], [0, 6], color='blue', linewidth=2)

# Annotate the magnitudes of the vectors using Pythagorean theorem
ax.text(1.7, 3.8, r'$\sqrt{4^2+7^2} = 8.06$', fontsize=16, rotation=np.arctan(7/4)*180/np.pi)
ax.text(5, 4.1, r'$\sqrt{8^2+6^2} = 10$', fontsize=16, rotation=np.arctan(6/8)*180/np.pi)

# Show the plot
plt.show()
```

# <font face="gotham"> Cộng, trừ, nhân véc tơ

**1. Cộng Vector**

Di chuyển theo vector $\mathbf{a}$, từ điểm cuối a, tiếp tục di chuyển theo vector $\mathbf{b}$. Vector tổng $\mathbf{a} + \mathbf{b}$ chính là "đoạn đường" từ điểm khởi đầu của $\mathbf{a}$ đến điểm cuối của $\mathbf{b}$.

*   **Hình học:** Dùng quy tắc hình bình hành hoặc quy tắc tam giác (nối đuôi-đầu).
*   **Đại số:** Nếu $\mathbf{a} = (x_1, y_1)$ và $\mathbf{b} = (x_2, y_2)$, thì $\mathbf{a} + \mathbf{b} = (x_1 + x_2, y_1 + y_2)$.

**2. Trừ Vector**

Phép trừ vector $\mathbf{a} - \mathbf{b}$ có thể hình dung như phép cộng vector $\mathbf{a}$ với vector đối của $\mathbf{b}$ (vector có cùng độ lớn nhưng ngược chiều với $\mathbf{b}$). 

*   **Hình học:** Nối điểm cuối của vector trừ (b) đến điểm cuối của vector bị trừ (a).
*   **Đại số:** Nếu $\mathbf{a} = (x_1, y_1)$ và $\mathbf{b} = (x_2, y_2)$, thì $\mathbf{a} - \mathbf{b} = (x_1 - x_2, y_1 - y_2)$.

**3. Nhân Vector với số vô hướng**

Khi vector $\mathbf{a}$ được nhân với một số vô hướng (scalar) $k$, vector mới $k\mathbf{a}$ sẽ có độ lớn thay đổi $k$ lần so với vector ban đầu. Nếu $k$ dương, hướng của vector không đổi; nếu $k$ âm, hướng của vector sẽ đảo ngược.

*   **Hình học:** Kéo dài hoặc rút ngắn vector theo tỉ lệ $k$, hoặc đảo chiều nếu $k < 0$.
*   **Đại số:** Nếu $\mathbf{a} = (x, y)$ và $k$ là một số vô hướng, thì $k\mathbf{a} = (kx, ky)$.

```python
fig, ax = plt.subplots(figsize = (9,9))

vec = np.array([[[0, 0, 4, 7]],
               [[0, 0, 8, 4]],
               [[0, 0, 12, 11]],
               [[4, 7, 8, 4]],
               [[8, 4, 4, 7]]])
color = ['r','b','g','b','r']

for i in range(vec.shape[0]):
    X,Y,U,V = zip(*vec[i,:,:])
    ax.quiver(X, Y, U, V, angles='xy', scale_units='xy', color = color[i], scale=1, alpha = .6)

ax.set_xlim([0, 15])
ax.set_ylim([0, 15])
ax.set_xlabel('Trục x', fontsize =16)
ax.set_ylabel('Trục y', fontsize =16)
ax.grid()

for i in range(3):
    ax.text(x = vec[i,0,2], y = vec[i,0,3], s = '(%.0d, %.0d)' %(vec[i,0,2],vec[i,0,3]), fontsize = 16)

ax.text(x= vec[0,0,2]/2, y = vec[0,0,3]/2, s= '$u$', fontsize = 16)
ax.text(x= 8, y = 9, s= '$v$', fontsize = 16)
ax.text(x= 6, y = 5.5, s= '$u+v$', fontsize = 16)

ax.set_title('Cộng véc tơ', size = 18)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(9, 9))

vec = np.array([[[0, 0, 4, 7]],
               [[0, 0, 8, 4]],
               [[0, 0, -8, -4]],
               [[0, 0, -4, 3]],
               [[-8, -4, 4, 7]],
               [[4, 7, -8, -4]],
               [[8, 4, -4, 3]]])
color = ['r','b','b','g','r','b','g']

for i in range(vec.shape[0]):
    X,Y,U,V = zip(*vec[i,:,:])
    ax.quiver(X, Y, U, V, angles='xy', scale_units='xy', color = color[i], scale=1, alpha = .6)

for i in range(4):
    ax.text(x = vec[i,0,2], y = vec[i,0,3], s = '(%.0d, %.0d)' %(vec[i,0,2],vec[i,0,3]), fontsize = 16)

s = ['$u$','$v$','$-v$','$u-v$']
for i in range(4):
    ax.text(x = vec[i,0, 2]/2, y = vec[i,0, 3]/2, s = s[i], fontsize = 16)
ax.text(x= 5.8 , y = 5.8, s= '$u-v$', fontsize = 16)

ax.set_xlim([-9, 9])
ax.set_ylim([-9, 9])
ax.set_xlabel('Trục x', fontsize =16)
ax.set_ylabel('Trục y', fontsize =16)
ax.grid()
ax.set_title('Trừ vector', size = 18)
plt.show()
```
